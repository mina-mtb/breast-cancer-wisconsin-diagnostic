# 4️⃣ در پزشکی کدام اشتباه خطرناک‌تر است؟

دو نوع اشتباه داریم:

### نوع 1 — False Positive

به یک فرد سالم بگویی سرطان دارد.

نتیجه:

* استرس
* آزمایش بیشتر
* اما زنده می‌ماند

### نوع 2 — False Negative  🚨

به یک فرد سرطانی بگویی سالم است.

نتیجه:

* درمان نمی‌شود
* سرطان پیشرفت می‌کند
* ممکن است بمیرد

پس:

> بدترین اتفاق برای یک مدل پزشکی = از دست دادن بیمار سرطانی

---

# 5️⃣ اینجاست که Recall مهم می‌شود

Recall (یا Sensitivity) یعنی:

[
Recall = \frac{سرطان‌هایی\ که\ پیدا\ کردیم}{کل\ سرطان‌های\ واقعی}
]

یعنی:

> از بین بیماران سرطانی، چند نفر را توانستیم کشف کنیم؟

---

### مثال

فرض کن 212 بیمار سرطانی داریم.

مدل A:

* فقط 40 نفر را پیدا کرد

[
Recall = 40/212 = 19%
]

مدل B:

* 190 نفر را پیدا کرد

[
Recall = 190/212 = 89%
]

حتی اگر مدل B چند سالم را اشتباه بگوید،
در پزشکی **خیلی بهتر** است.

---

# نتیجه‌ای که استاد می‌خواهد تو بفهمی

در سیستم‌های تشخیصی:

❌ هدف اصلی: بیشترین Accuracy
✅ هدف اصلی: از دست ندادن بیمار واقعی

پس:

> ما مدل را بر اساس **Recall برای Malignant** قضاوت می‌کنیم، نه Accuracy.

این دقیقاً دلیل وجود ماموگرافی، آزمایش‌های اضافی و biopsies در دنیای واقعی است.

---

اگر دوست داشتی، قدم بعدی می‌توانم با confusion matrix خودت نشان بدهم کدام خانه دقیقاً «بیمار از دست رفته» است — آن لحظه معمولاً این مفهوم کاملاً جا می‌افتد.


````python
y = (df["diagnosis"] == "M").astype(int)   # 1 = cancer
X = df.drop(columns=["id", "diagnosis"])
````

گام 2 — تعریف "غیرعادی" (Abnormal)

ما پزشک نیستیم، پس از آمار استفاده می‌کنیم:

قانون قابل دفاع علمی:

مقادیر خیلی بزرگ (outliers) نشانه سرطان هستند.

یک روش استاندارد:

بالاتر از صدک 90 (Top 10%)

یعنی:
بزرگ‌ترین 10٪ سلول‌ها → مشکوک.

حالا 4 قانون پزشکی را می‌سازیم

ما از هر مفهوم پزشکی یک نماینده انتخاب می‌کنیم:

| مفهوم پزشکی             | ویژگی دیتاست           |
| ----------------------- | ---------------------- |
| اندازه (size)           | `radius_worst`         |
| شکل (shape)             | `concave_points_worst` |
| بافت (texture)          | `texture_worst`        |
| ناهمگنی (heterogeneity) | `radius_se`            |


````python
## محاسبه آستانه برای هر ویژگی
size_thr   = X["radius_worst"].quantile(0.90)
shape_thr  = X["concave_points_worst"].quantile(0.90)
texture_thr = X["texture_worst"].quantile(0.90)
hetero_thr  = X["radius_se"].quantile(0.90)

size_thr, shape_thr, texture_thr, hetero_thr
````

````python
## تعیین نمونه‌های غیرطبیعی بر اساس آستانه‌ها
size_abnormal   = X["radius_worst"] > size_thr
shape_abnormal  = X["concave_points_worst"] > shape_thr
texture_abnormal = X["texture_worst"] > texture_thr
hetero_abnormal  = X["radius_se"] > hetero_thr
````

````python
## ترکیب قوانین برای پیش‌بینی نهایی
rule_pred = (
    size_abnormal |
    shape_abnormal |
    texture_abnormal |
    hetero_abnormal
).astype(int)
````

````python

from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y, rule_pred))
print(classification_report(y, rule_pred))
````

# حالا می‌رویم سر مدل دوم (ML واقعی)




````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
````

````python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=300, random_state=42)
rf.fit(X_train, y_train)
````

````python
rf_pred = rf.predict(X_test)

from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y_test, rf_pred))
print(classification_report(y_test, rf_pred))
````
